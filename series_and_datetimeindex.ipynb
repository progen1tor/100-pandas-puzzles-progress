## Series and DatetimeIndex

```python
import pandas as pd 
import numpy as np 
```

**33.** Create a DatetimeIndex that contains each business day of 2015 and use it to index a Series of random numbers. Let's call this Series `s`.

```python
dt_idx = pd.bdate_range(start='2015-01-01', end='2015-12-31')
s = pd.Series(np.random.randint(1, 1000, 261), index=dt_idx)

display(s.head())
display(s.tail())
```

**34.** Find the sum of the values in `s` for every Wednesday.

```python
s.loc[s.index.day_of_week == 2].sum()
```

**35.** For each calendar month in `s`, find the mean of values.

```python
s.groupby(lambda d: d.month).mean()
```

**36.** For each group of four consecutive calendar months in `s`, find the date on which the highest value occurred.

```python
s.groupby(lambda d: (d.month - 1) // 4).idxmax()
```

**37.** Create a DateTimeIndex consisting of the third Thursday in each month for the years 2015 and 2016.

```python
dt_idx2 = pd.DatetimeIndex(pd.date_range(start='2015-01-01', end='2016-12-31', freq='D'))
df = pd.DataFrame(np.random.randint(1, 1000, 731), index=dt_idx2)
df = df.loc[df.index.day_of_week == 3]
df['th_num'] = df.groupby(pd.Grouper(freq='ME')).cumcount() + 1 
display(df.loc[df.th_num == 3].index)

# while checking my solution it turned out that we can create datetimeindex w/special frec xd
display(pd.date_range(start='2015-01-01', end='2016-12-31', freq='WOM-3THU'))
```
